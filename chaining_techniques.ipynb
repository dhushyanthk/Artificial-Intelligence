```python
facts = {
    "Human(Socrates)"
}
rules = {
    "Mortal(Socrates)": "Human(Socrates)"
}
def backward_chain(goal):
    print(f"\nTrying to prove: {goal}")
    if goal in facts:
        print(f"{goal} is a FACT.")
        return True
    if goal in rules:
        premise = rules[goal]
        print(f"Need to prove: {premise}")
        if backward_chain(premise):
            print(f"Rule applied:")
            print(f"{premise} --> {goal}")
            return True
    print(f"Cannot prove {goal}")
    return False
goal = "Mortal(Socrates)"
print("Backward Chaining")
if backward_chain(goal):
    print("\nGoal PROVED")
else:
    print("\nGoal NOT PROVED")
```

```python
facts = {"tom is a cat"}
rules = [
    ("is a mammal", "is a cat"), # If X is a cat, then X is a mammal
    ("is an animal", "is a mammal") # If X is a mammal, then X is an animal
]

# Simple forward chaining
new_facts_derived = True
while new_facts_derived:
    new_facts_derived = False
    for current_fact in list(facts): # Iterate over a copy to allow modification
        for consequent, antecedent in rules:
            # A very simple pattern matching for 'X is Y'
            if current_fact.endswith(antecedent):
                # Extract X from 'X is antecedent'
                subject = current_fact[:-len(antecedent)].strip()
                inferred_fact = f"{subject} {consequent}"
                if inferred_fact not in facts:
                    facts.add(inferred_fact)
                    new_facts_derived = True

print("Knowledge Base:")
for fact in facts:
    print(fact)
```

## First-Order Logic Representation

First-Order Logic (FOL) allows for a more expressive way to represent knowledge by using predicates, functions, constants, variables, and quantifiers. Unlike propositional logic, FOL can express relationships between individuals and generalize across them.

Here's how your facts can be represented using predicates:

```python
# Representing facts in First-Order Logic
fol_facts = {
    "Teacher(anita)",
    "Student(Arjun)",
    "Student(meena)",
    "WritesExam(arjun)",
    "WritesExam(meena)"
}

print("First-Order Logic Facts:")
for fact in fol_facts:
    print(fact)
```

### Example of a Rule in First-Order Logic

A rule in FOL typically involves implications and can use variables to generalize. For instance, if we wanted to express that 'Any student who writes an exam is eligible for a grade', we could write it as:

$\forall X \text{ (Student(X) } \land \text{ WritesExam(X) } \implies \text{ EligibleForGrade(X))}$

In a Python context, implementing a full FOL inference engine is complex, but the rules would logically follow this structure. We can represent such a rule in a simplified Python dictionary for illustrative purposes, where keys are conclusions and values are premises (which could be combined using 'AND').

```python
# Example of a rule structure for illustrative purposes (not a full FOL engine)
fol_rules = {
    # Conclusion: EligibleForGrade(X)
    # Premises: Student(X) AND WritesExam(X)
    "EligibleForGrade": {"premises": ["Student", "WritesExam"], "variables": ["X"]}
}

print("\nFirst-Order Logic Rule Example:")
for rule_name, rule_data in fol_rules.items():
    premises = ' AND '.join([f"{p}({rule_data['variables'][0]})" for p in rule_data['premises']])
    conclusion = f"{rule_name}({rule_data['variables'][0]})"
    print(f"For all {rule_data['variables'][0]}: ({premises}) => {conclusion}")
```

Based on your input, here's a program that can process your facts and rules. I've interpreted your rule `"writeexam":student` as a forward chaining rule: if `student(X)` is a fact, then we can infer `writeexam(X)`.

```python
import re

# User-provided input for facts and rules
user_facts_str = "teacher(anita) student(arjun) student(meena)"
user_rules_dict = {"writeexam": "student"}

# 1. Parse the facts
# Extract individual facts from the string
facts = set(re.findall(r'\w+\([^)]+\)', user_facts_str))

# 2. Represent the rules in a structured format for forward chaining
# rule_name: (premise_predicate, conclusion_predicate)
# For 'writeexam': 'student' we assume: if student(X) then writeexam(X)
parsed_rules = []
for conclusion_pred, premise_pred in user_rules_dict.items():
    parsed_rules.append((premise_pred, conclusion_pred))

print("Initial Facts:")
for fact in sorted(list(facts)):
    print(fact)

print("\nRules being applied:")
for premise, conclusion in parsed_rules:
    print(f"IF {premise}(X) THEN {conclusion}(X)")

# 3. Implement Forward Chaining
new_facts_derived = True
iteration = 0
while new_facts_derived:
    new_facts_derived = False
    iteration += 1
    print(f"\n--- Applying Rules (Iteration {iteration}) ---")
    current_facts_copy = list(facts) # Iterate over a copy to allow modification

    for current_fact in current_facts_copy:
        # Check for rules that can be applied
        for premise_pred, conclusion_pred in parsed_rules:
            # Simple pattern matching: check if current_fact matches a premise predicate
            if current_fact.startswith(premise_pred + '(') and current_fact.endswith(')'):
                # Extract the argument (e.g., 'arjun' from 'student(arjun)')
                match = re.match(rf'{premise_pred}\(([^)]+)\)', current_fact)
                if match:
                    argument = match.group(1)
                    inferred_fact = f"{conclusion_pred}({argument})"

                    if inferred_fact not in facts:
                        facts.add(inferred_fact)
                        new_facts_derived = True
                        print(f"Derived new fact: {inferred_fact} from {current_fact}")

# 4. Display Final Knowledge Base
print("\nFinal Knowledge Base:")
for fact in sorted(list(facts)):
    print(fact)
```
